```python
#Sel tes: memastikan perintah HDFS dapat dipanggil dari Jupyter
!hdfs version
```

```python
# Membuat direktori kerja pribadi di HDFS
# HDFS memiliki struktur folder sendiri, terpisah dari folder Linux biasa
!hdfs dfs -mkdir -p /user/mazedev/praktikum

# Menampilkan isi direktori root HDFS untuk verifikasi
!hdfs dfs -ls /user/mazedev
```

```python
# Membuat berkas teks sederhana di disk lokal (bukan HDFS) sebagai contoh
with open("contoh_lokal.txt", "w") as f:
    f.write("Halo dari Praktikum Big Data!\n")
    f.write("Berkas ini awalnya tersimpan di disk lokal.\n")

print("Berkas lokal berhasil dibuat.")
!ls -la contoh_lokal.txt
```

```python
# Upload (put) berkas lokal ke dalam HDFS
!hdfs dfs -put contoh_lokal.txt /user/mazedev/praktikum/

# Verifikasi berkas sudah ada di HDFS
!hdfs dfs -ls /user/mazedev/praktikum
```

```python
!hdfs dfs -cat /user/mazedev/praktikum/contoh_lokal.txt
```

```python
# Mengunduh (get) berkas dari HDFS kembali ke disk lokal dengan nama berbeda
!hdfs dfs -get /user/mazedev/praktikum/contoh_lokal.txt hasil_unduh_dari_hdfs.txt

# Verifikasi isinya sama
!cat hasil_unduh_dari_hdfs.txt
```

```python
# -du: menampilkan ukuran penggunaan disk di HDFS
!hdfs dfs -du -h /user/mazedev/praktikum

# -count: menampilkan jumlah direktori, berkas, dan total ukuran
!hdfs dfs -count /user/mazedev/praktikum
```

```python
# Menghapus satu berkas di HDFS
!hdfs dfs -rm /user/mazedev/praktikum/contoh_lokal.txt

# Verifikasi berkas sudah terhapus (direktori seharusnya kosong sekarang)
!hdfs dfs -ls /user/mazedev/praktikum
```

```python
#Latihan Mandiri
# Buatlah direktori baru di HDFS dengan path /user/mahasiswa/latihan3
!hdfs dfs -mkdir -p /user/mazedev/latihan3

# Menampilkan isi direktori root HDFS untuk verifikasi
!hdfs dfs -ls /user/mazedev
```

```python
# Buatlah sebuah berkas teks lokal berisi biodata singkat kalian (nama, NIM, hobi — minimal 3 baris), 
# lalu upload ke direktori `/user/mahasiswa/latihan3` yang baru dibuat pada Soal 1.
with open("biodata.txt", "w") as f:
    f.write("Nama\t: Emannuel Henji Pratama Putra\n")
    f.write("NIM\t: 2505060012\n")
    f.write("Hobi\t: Ngoding\n")

print("Berkas lokal berhasil dibuat.")
!ls -la biodata.txt
```

```python
# Tampilkan isi berkas
!hdfs dfs -cat /user/mazedev/latihan3/biodata.txt
```
