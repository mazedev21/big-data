## 1. Persiapan Dataset & SparkSession

Membuat dataset simulasi `tugas11_penjualan.csv` sesuai instruksi modul, kemudian menginisialisasi `SparkSession` dan membaca file CSV tersebut ke dalam Spark DataFrame.

```python
import numpy as np
import pandas as pd

np.random.seed(111)
n = 2500
menu_list = ["Kopi Susu", "Americano", "Cappuccino", "Roti Bakar", "Kentang Goreng", "Es Teh"]
cabang_list = ["Magelang", "Yogyakarta", "Semarang"]
waktu_list = ["Pagi", "Siang", "Sore", "Malam"]

data_tugas11 = pd.DataFrame({
    "nama_menu": np.random.choice(menu_list, size=n),
    "cabang": np.random.choice(cabang_list, size=n),
    "waktu_pesan": np.random.choice(waktu_list, size=n),
    "qty": np.random.randint(1, 6, size=n),
    "harga_satuan": np.random.choice([15000, 18000, 20000, 22000, 25000, 30000], size=n),
})
data_tugas11["total_penjualan"] = data_tugas11["qty"] * data_tugas11["harga_satuan"] + np.random.normal(0, 8000, size=n)
data_tugas11.to_csv("tugas11_penjualan.csv", index=False)

print(f"Dataset berhasil dibuat: {len(data_tugas11)} baris")
print(data_tugas11.head())
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Tugas11-MLlibRegresi") \
    .master("local[*]") \
    .getOrCreate()
spark.sparkContext.setLogLevel("ERROR")
spark.conf.set("spark.sql.shuffle.partitions", "4")

print("SparkSession siap. Versi Spark:", spark.version)

df = spark.read.csv("tugas11_penjualan.csv", header=True, inferSchema=True)
print("Total baris:", df.count())
df.printSchema()
df.show(5)
```

---
## A. Feature Engineering

Mengubah kolom kategorikal (`nama_menu`, `cabang`, dan `waktu_pesan`) menjadi indeks numerik menggunakan `StringIndexer` (menyimpan objek model hasil `fit()` agar dapat digunakan kembali pada Bagian E), lalu menggabungkan `qty`, `harga_satuan`, dan ketiga kolom hasil indexing ke dalam kolom `features` menggunakan `VectorAssembler`.

```python
from pyspark.ml.feature import StringIndexer, VectorAssembler

indexer_menu = StringIndexer(inputCol="nama_menu", outputCol="nama_menu_idx")
indexer_cabang = StringIndexer(inputCol="cabang", outputCol="cabang_idx")
indexer_waktu = StringIndexer(inputCol="waktu_pesan", outputCol="waktu_pesan_idx")

# Simpan model hasil fit() agar dapat dipakai ulang saat memprediksi data baru di Bagian E
model_indexer_menu = indexer_menu.fit(df)
df_indexed = model_indexer_menu.transform(df)

model_indexer_cabang = indexer_cabang.fit(df_indexed)
df_indexed = model_indexer_cabang.transform(df_indexed)

model_indexer_waktu = indexer_waktu.fit(df_indexed)
df_indexed = model_indexer_waktu.transform(df_indexed)

df_indexed.select(
    "nama_menu", "nama_menu_idx",
    "cabang", "cabang_idx",
    "waktu_pesan", "waktu_pesan_idx"
).show(8)

assembler = VectorAssembler(
    inputCols=["qty", "harga_satuan", "nama_menu_idx", "cabang_idx", "waktu_pesan_idx"],
    outputCol="features"
)
df_final = assembler.transform(df_indexed)
df_final.select("features", "total_penjualan").show(5, truncate=False)
```

---
## B. Train/Test Split & Pelatihan Model

Membagi data menjadi 75% training dan 25% testing dengan `seed=101`, kemudian melatih model `LinearRegression` dengan `labelCol="total_penjualan"`.

```python
from pyspark.ml.regression import LinearRegression

train, test = df_final.randomSplit([0.75, 0.25], seed=101)
print(f"Jumlah data training : {train.count()}")
print(f"Jumlah data testing  : {test.count()}")

lr = LinearRegression(featuresCol="features", labelCol="total_penjualan")
model = lr.fit(train)

print("Model berhasil dilatih!")
print("Koefisien tiap fitur:", model.coefficients)
print("Intercept:", model.intercept)
```

---
## C. Evaluasi Model

Menghitung dan menampilkan **RMSE** serta **R²** model utama pada data testing.

```python
from pyspark.ml.evaluation import RegressionEvaluator

predictions = model.transform(test)
predictions.select(
    "nama_menu", "cabang", "waktu_pesan", "qty", "harga_satuan", "total_penjualan", "prediction"
).show(10)

evaluator_rmse = RegressionEvaluator(labelCol="total_penjualan", predictionCol="prediction", metricName="rmse")
rmse = evaluator_rmse.evaluate(predictions)

evaluator_r2 = RegressionEvaluator(labelCol="total_penjualan", predictionCol="prediction", metricName="r2")
r2 = evaluator_r2.evaluate(predictions)

print(f"RMSE : Rp {rmse:,.0f}")
print(f"R²   : {r2:.4f}  ({r2*100:.1f}% variasi data berhasil dijelaskan model)")
```

**Interpretasi Model (Bagian C):**
*(Akan diperbarui sesuai hasil eksekusi)*

---
## D. Eksperimen Perbandingan Fitur

Melatih model kedua yang **hanya** menggunakan fitur `qty` dan `harga_satuan` (tanpa informasi `nama_menu`, `cabang`, dan `waktu_pesan`), lalu membandingkan nilai **R²** dan **RMSE** dengan model utama pada Bagian C.

```python
assembler_sederhana = VectorAssembler(
    inputCols=["qty", "harga_satuan"],
    outputCol="features_sederhana"
)
df_final_2 = assembler_sederhana.transform(df)

train_2, test_2 = df_final_2.randomSplit([0.75, 0.25], seed=101)

lr_2 = LinearRegression(featuresCol="features_sederhana", labelCol="total_penjualan")
model_2 = lr_2.fit(train_2)

print("Model kedua berhasil dilatih!")
print("Koefisien (qty, harga_satuan):", model_2.coefficients)
print("Intercept:", model_2.intercept)

predictions_2 = model_2.transform(test_2)
rmse_2 = RegressionEvaluator(labelCol="total_penjualan", predictionCol="prediction", metricName="rmse").evaluate(predictions_2)
r2_2 = RegressionEvaluator(labelCol="total_penjualan", predictionCol="prediction", metricName="r2").evaluate(predictions_2)

print(f"Model 1 (5 fitur) -> RMSE: Rp {rmse:,.0f} | R²: {r2:.4f}")
print(f"Model 2 (2 fitur) -> RMSE: Rp {rmse_2:,.0f} | R²: {r2_2:.4f}")
print(f"Selisih R²        -> {r2 - r2_2:+.6f}")
```

**Kesimpulan Perbandingan (Bagian D):**
*(Akan diperbarui sesuai hasil eksekusi)*

---
## E. Prediksi Skenario Baru

Membuat DataFrame berisi 3 skenario pesanan dummy, memprosesnya menggunakan model `StringIndexer` yang **sama** (hasil `fit()` di Bagian A) dan `VectorAssembler` yang sama, lalu menampilkan hasil prediksi `total_penjualan`.

```python
skenario_baru = spark.createDataFrame([
    ("Kopi Susu", "Magelang", "Pagi", 3, 20000),
    ("Cappuccino", "Yogyakarta", "Sore", 2, 25000),
    ("Roti Bakar", "Semarang", "Malam", 4, 18000),
], ["nama_menu", "cabang", "waktu_pesan", "qty", "harga_satuan"])

skenario_indexed = model_indexer_menu.transform(skenario_baru)
skenario_indexed = model_indexer_cabang.transform(skenario_indexed)
skenario_indexed = model_indexer_waktu.transform(skenario_indexed)

skenario_final = assembler.transform(skenario_indexed)
prediksi_skenario = model.transform(skenario_final)

prediksi_skenario.select(
    "nama_menu", "cabang", "waktu_pesan", "qty", "harga_satuan", "prediction"
).show(truncate=False)
```

```python
spark.stop()
print("SparkSession ditutup.")
```
